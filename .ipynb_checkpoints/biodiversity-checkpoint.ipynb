##### Cleaning the rows with same scientific_name

```python
# Split the common_names column so each common name appears in a separate row
new_rows = []
for _, row in species.iterrows():
    common_names = row['common_names'].split(',')
    for name in common_names:
        new_row = row.copy()  
        new_row['common_names'] = name.strip()
        new_rows.append(new_row)

species = pd.DataFrame(new_rows)
```

```python
# Check duplicates
species.duplicated().sum()
```

```python
# Check which scientific names appear more than once in the dataset
species['scientific_name'].value_counts()[species['scientific_name'].value_counts() > 1]
```

```python
# Check if any scientific name has more than one conservation status
status_counts = species.groupby('scientific_name')['conservation_status'].nunique()
status_counts[status_counts > 1]
```

```python
# Inspect the rows for Canis lupus to understand duplicated scientific names and conservation status
species[species['scientific_name'] == 'Canis lupus']
```

```python
# Inspect the rows for Oncorhynchus mykiss to understand duplicated scientific names and conservation status
species[species['scientific_name'] == 'Oncorhynchus mykiss']
```

I found that some scientific names have more than one conservation status. To avoid duplicating observation records after merging, I will keep only one row per scientific name. When a species has conflicting conservation statuses, I will keep the status with the highest conservation concern, because it is safer to overestimate conservation risk than to underestimate it.

```python
# Creates a priority order of conservation concern
conservation_priority= {
    'No Intervention' : 0,
    'Species of Concern' : 1,
    'In Recovery' : 2,
    'Threatened' : 3,
    'Endangered' : 4
}

species['conservation_priority'] = species['conservation_status'].map(conservation_priority)
```

```python
# Sort species so that each scientific name is grouped together
species = species.sort_values(by=['scientific_name', 'conservation_priority'])
species.head()
```

```python
# Create one row per scientific name by combining its distinct common names and keeping the row with the highest conservation priority.

clean_species = []

for scientific_name in species['scientific_name'].unique():

    one_species = species[species['scientific_name'] == scientific_name]

    common_names = []
    
    for name in one_species['common_names']:
        if name not in common_names:
            common_names.append(name)

    final_row = one_species.iloc[-1].copy()
    
    final_row['common_names'] = ', '.join(common_names)
    
    clean_species.append(final_row)

species = pd.DataFrame(clean_species)
```

```python
# Reorder the columns and remove the conservation_priority column from the final cleaned dataset
species = species[['category', 'scientific_name', 'common_names', 'conservation_status']]
species.head()
```

### Analysis

```python
# Explore the categories distribution
categories_distribution = species.groupby('category').size()
categories_distribution
```

```python
# Explore the conservation distribution
conservation_distribution = species.groupby('conservation_status')['scientific_name'].nunique()
conservation_distribution
```

```python
# Explore observations by park
total_observations_by_park = observations.groupby('park_name')['observations'].sum()
total_observations_by_park
```

```python
# Explore observations by park
species_by_park = observations.groupby('park_name')['scientific_name'].nunique()
species_by_park
```

```python
# Filter species with a conservation status other than 'No Intervention'
threatened_data = species[species['conservation_status'] != 'No Intervention']

# Group by 'category' and 'conservation_status', and count unique species
conservation_by_category = (
    threatened_data.groupby(['category', 'conservation_status'])
    ['scientific_name'].nunique().unstack())


conservation_by_category
```

```python
# Chi-square test
chi2_cat, p_val, dof_cat, expected_cat = chi2_contingency(conservation_by_category.fillna(0))

print(f'Conservation status by Category p-value: {p_val:.2e}')
```

**With a p-value of 2.52e-05, which is below the significance level of 0.05, we reject the null hypothesis. This suggests that conservation status and category are significantly associated.**

```python
# Define colors for each conservation status
palette = sns.color_palette('deep', n_colors = 4)

# Plot a stacked bar chart of conservation statuses by species category
plt.figure(figsize=(16,10))
conservation_by_category.plot(kind = 'bar', stacked = True, color = palette)
plt.title('Conservation Status by Category - Number of Species')
plt.xlabel('Category')
plt.ylabel('Number of Species')
plt.xticks(rotation=45)
plt.legend(title = 'Conservation Status', bbox_to_anchor=(1.5, 0.5))

plt.show()
```

```python
# Merge observations with threatened species information
threatened_merged_data = observations.merge(threatened_data, on = 'scientific_name')

# Calculate total observations by category and conservation status
obs_by_cat_conservation = (
    threatened_merged_data.groupby(['category', 'conservation_status'])['observations'].sum().unstack())

obs_by_cat_conservation
```

```python
plt.figure(figsize = (8,6))

# Create a heatmap showing total observations by category and conservation status
sns.heatmap(obs_by_cat_conservation.fillna(0), annot = True, fmt='.0f', linewidth=0.8, cmap = 'Reds')

plt.title('Observations by Conservation Status and Category')
plt.xlabel('Conservation Status', labelpad = 15)
plt.ylabel('Category', labelpad = 15)

plt.show()
```

```python
# Merge the species data with the observations data using the scientific name
merged_data = species.merge(observations, on = 'scientific_name').sort_values('observations', ascending = False).reset_index(drop = True)
merged_data.head()
```

```python
# Group the merged data by species category and park name and calculates the number of observations
obs_cat_park = (merged_data.groupby(['category', 'park_name'])['observations'].sum().unstack())
obs_cat_park
```

```python
plt.figure(figsize = (8,6))
sns.heatmap(obs_cat_park, annot = True, fmt='.0f', linewidth=0.8, cmap = 'Greens')

plt.title('Observations by Category and Park', pad = 20)
plt.xlabel('Park', labelpad = 10)
plt.xticks(rotation = 45)
plt.ylabel('Category', labelpad = 15)

plt.show()
```

#### Conservation Status

```python
# If the conservation status is 'No Intervention', classify the species as 'Unprotected'. Otherwise, classify the species as 'Protected'
species['protection_status'] = np.where(species['conservation_status'] == 'No Intervention', 'Unprotected', 'Protected')
species.head()                                       
```

```python
# Count the number of unique species in each category by protection status
conservation = (species.groupby(['category', 'protection_status'])['scientific_name'].nunique().unstack().reset_index())

conservation.columns = (['Category', 'Protected', 'Unprotected'])
conservation
```

```python
data = conservation[['Protected', 'Unprotected']].values 

# chi_square test
chi2_cat, pval, dof_cat, expected_cat = chi2_contingency(data)

print(f'p value: {pval}')
```

**Since the p-value is much smaller than 0.05, we reject the null hypothesis. This suggests that species category and protection status are significantly associated. The proportion of protected and unprotected species is not the same across all categories.**

```python
fig, ax = plt.subplots(1, 2, figsize=(14, 6))

#'Protected' plot
ax[0].bar(conservation['Category'], conservation['Protected'], color = '#6a994e')
ax[0].set_title('Protected Species by Category')
ax[0].set_xlabel('Category')
ax[0].set_ylabel('Number of Species')
ax[0].tick_params(axis = 'x', rotation = 45)

#'Unprotected' plot
ax[1].bar(conservation['Category'], conservation['Unprotected'], color = '#c1121f')
ax[1].set_title('Unprotected Species by Category')
ax[1].set_xlabel('Category')
ax[1].set_ylabel('Number of Species')
ax[1].tick_params(axis = 'x', rotation = 45)

plt.tight_layout()
plt.show()
```

### Study of species in the bird category


```python
# Creates a function to remove punctuation using regular expressions
def remove_punctuation(text):
    return re.sub(r'[^\w\s]', '', text)
```

```python
# Filter the species dataframe to keep only rows where the category is 'Bird'
bird_species = species[species['category'] == 'Bird'].reset_index(drop = True)

# Clean the common names by removing punctuation
bird_common_names = (bird_species['common_names'].apply(remove_punctuation).str.split().tolist())
```

```python
# Remove repeated words from each bird common name
clean_bird_names = []

for name in bird_common_names:
    unique_words = []
    for word in name:
        if word not in unique_words:
            unique_words.append(word)
    clean_bird_names.append(unique_words)
```

```python
# Combine all words from the cleaned bird names into one single list
bird_names = []

for name in clean_bird_names:
    for word in name:
        bird_names.append(word)
```

```python
# Count the frequency of each word in the bird common names and create a sorted dataframe with the results.
bird_names_series = pd.Series(bird_names)

names_counted = (bird_names_series.value_counts().reset_index())

names_counted.columns = ['Name', 'Count']

birds_df = names_counted.sort_values('Count', ascending = False)
birds_df.head()
```

```python
# Identify and filter bird species whose common names contain 'Warbler'

bird_species['is_warbler'] = bird_species['common_names'].str.contains('Warbler')
warbler_birds = bird_species[bird_species['is_warbler']]
```

```python
# Combine warbler species with their observations and sort by highest number of observations
warbler_observations = (
    warbler_birds.merge(observations, on = 'scientific_name')
    .sort_values('observations', ascending = False)
    .reset_index(drop = True))

warbler_observations.head()
```

```python
# Calculate total warbler observations by park and protection status
warbler_obs_by_park = (warbler_observations.groupby(['park_name', 'protection_status'])['observations'].sum().reset_index())
warbler_obs_by_park
```

```python
# Plot total warbler observations by park, separated by protection status
colors = {'Protected': '#38b000', 'Unprotected': '#d00000'}

plt.figure(figsize = (8,6))
sns.barplot(x = 'park_name', y = 'observations', hue = 'protection_status', data = warbler_obs_by_park, palette = colors )

plt.title('Observation of Warblers by park')
plt.legend(title = 'Protection Status')
plt.xlabel('Park', labelpad = 10)
plt.xticks(rotation = 45)
plt.ylabel('Observations', labelpad = 15)

plt.show()
```

### Conclusions

- The analysis showed that most species in the dataset are not currently under conservation intervention. Out of 5,541 species, 5,362 were classified as having “No Intervention”, while 179 species had some level of conservation status.

- When looking at conservation status by category, Birds had the highest number of protected species, followed by Mammals and Vascular Plants. However, relative to the size of each category, Mammals and Birds had a higher proportion of protected species compared with most other groups.

- The chi-square tests showed statistically significant relationships between species category and conservation status. This means that conservation status is not evenly distributed across all species categories. Some categories are more likely to include protected species than others.

- Vascular Plants were the most common category in the dataset, representing 4,262 species. They also had the highest number of observations across the national parks. Among the parks, Yellowstone National Park had the highest total number of observations, followed by Yosemite National Park, Bryce National Park, and Great Smoky Mountains National Park.

- In the bird-specific analysis, “Warbler” was the most frequent word found in bird common names. Warbler observations were highest in Yellowstone National Park, and most of these observations came from unprotected species.

- Overall, the results suggest that conservation efforts are concentrated in specific species categories, especially Birds and Mammals. The findings also show that observation numbers vary strongly by park, with Yellowstone having the largest number of recorded observations.
